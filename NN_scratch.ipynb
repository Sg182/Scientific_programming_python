```python
# A BASIC MULTILEVEL NN 
import torch
import torch.nn as nn

torch.manual_seed(42)
X = torch.tensor([[2.0,3.0]])

class NeuralNetwork(nn.Module):

    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(2, 3)    # 2 inputs   3 neurons (outputs)
        self.layer2 = nn.Linear(3, 2)    # 3 inputs   2 neurons (outputs)
        self.layer3 = nn.Linear(2, 1)    # 2 inputs   1 neuron
        self.relu = nn.ReLU()

    def forward(self, x):

        x = self.relu(self.layer1(x))
        x = self.relu(self.layer2(x))
        x = self.layer3(x)

        return x
        
model = NeuralNetwork()

print("The predicted outcome is :",model.forward(X))
```

```python
# OPTIMIZING THE WEIGHTS OF A MULTILEVEL NN

torch.manual_seed(42)
X = torch.tensor([
    [2.0, 3.0],
    [1.0, 4.0],
    [3.0, 1.0]
])

y = torch.tensor([[5.0],[6.0],[4.0]])

class NeuralNetwork(nn.Module):

    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(2, 3)    # 2 inputs   3 neurons (outputs)
        self.layer2 = nn.Linear(3, 2)    # 3 inputs   2 neurons (outputs)
        self.layer3 = nn.Linear(2, 1)    # 2 inputs   1 neuron
        self.relu = nn.ReLU()

    def forward(self, x):

        x = self.relu(self.layer1(x))
        x = self.relu(self.layer2(x))
        x = self.layer3(x)

        return x


model = NeuralNetwork()
loss_fun = nn.MSELoss()
print("type of model ",type(model))
print("type of loss_fun ",type(loss_fun))
#INITIALIZE AN OPTIMIZER

optimizer = optim.SGD(model.parameters(),lr=0.01)

#Now start the optimization

for epoch in range(100):

    #FORWARD PROPAGATION
    prediction = model(X)

    #COMPUTE THE LOSS
    loss  = loss_fun(prediction,y)
    optimizer.zero_grad()

    #BACKPROPAGATION
    loss.backward()

    #UPDATE THE WEIGHTS AND BIAS
    optimizer.step()

 
print("\nOptimized parameters:")

'''for name, param in model.named_parameters():
    print(name)
    print(param.data)'''

print("FINAL WEIGHTS: ",model.layer3.weight.data)
print("FINAL BIAS: ",model.layer3.bias.data)
```
